```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step 2 : Read the Data**

```python
import pandas as pd

data = {
    "hours": [1, 2, 3, 4, 5],
    "marks": [50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
x = df[["hours"]].values
y = df[["marks"]].values
```

# **Model Fitting**

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
slope = model.coef_[0]
intercept = model.intercept_
```

```python
print(f"slope : {slope}")
print(f"intercept (b): {intercept}")
print(f"Equation of the regression line: y={slope[0]:.2f} x + {intercept[0]:.2f}")
```

```python
df['Predicted Marks'] = model.predict(x)
```

```python
df.head()
```

# **Visualizing**

```python
plt.scatter(df['hours'],df['marks'],color = 'brown',label = 'Actual Marks')
```

# **Plotting the Equation**

```python
plt.plot(df['hours'],df['Predicted Marks'],color = 'green',label = 'Regression Line ')
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.title('Hours vs Marks')
plt.legend()
plt.show()
```
